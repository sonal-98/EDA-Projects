### Exploratory Data Analysis of Agriculture Crop Production in India
#### Objective
The objective of this project is to analyse agricultural crop production in India. The analysis uses Python, Pandas, NumPy, Matplotlib, and Seaborn to clean, explore, visualise, and answer business questions from the dataset.

```python
# Importing libraries 

import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import matplotlib.pyplot as plt   # For visualization
import seaborn as sns   # for more advanced and attracting visuals
```

```python
# Reading the 3 CSV files needed to be used in this project

df1=pd.read_csv('/kaggle/input/datasets/srinivas1/agricuture-crops-production-in-india/datafile (1).csv')
df2=pd.read_csv('/kaggle/input/datasets/srinivas1/agricuture-crops-production-in-india/datafile (2).csv')
df3=pd.read_csv('/kaggle/input/datasets/srinivas1/agricuture-crops-production-in-india/datafile (3).csv')
```

### Business Questions

1. Which state have the highest cultivation cost?

```python
merge['Total_cost']=merge['Cultivation_Cost(A2+FL)']+merge['Cultivation_Cost(C2)']
merge['Total_cost']
```

```python
cost_per_state= merge.groupby('State')['Total_cost'].mean().sort_values(ascending=False)
cost_per_state
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x=cost_per_state.values,
    y=cost_per_state.index,
    palette='magma'
)

plt.title("Average cost of cultivation per state")

plt.xlabel("Cultivation Cost")

plt.ylabel("State")

plt.show()
```

2] Top 5 Season segments contributing to crop production?

```python
merge.groupby('Season/ duration in days')['Total_cost'].size().sort_values(ascending=False).head(5)
```
