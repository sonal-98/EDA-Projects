### Business questions

1. How many days before arrival do guests usually book?

```python
# We need to find the Average Lead time of all the bookings

avg=df['lead_time'].mean()
print(f'Average Lead time: {avg:.2f}')
```

```python
plt.figure(figsize=(10,5))
plt.hist(df['lead_time'], bins=40)
plt.title("Distribution of Booking Lead Time")
plt.xlabel("Lead time (Days)")
plt.ylabel("Number of Bookings")

plt.show()
```

***Higher average value, it means customers book more in advance***

2. What percentage of all bookings are cancelled??

```python
booking_canceled= (df['is_canceled'].mean())*100
print("Canceled bookings percentage:", round(booking_canceled,2))
```

***A lower percentage of bookings canceled,it means a higher number of successful bookings.***

3. Which customer segment generates the highest room revenue?

```python
df['TotalNights']=df['stays_in_weekend_nights']+df['stays_in_week_nights']
```

```python
df['total_rev']=df['adr']* df['TotalNights']
```

```python
rev=df.groupby('customer_type')['total_rev'].sum().sort_values(ascending=False)
rev
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    x=rev.index,
    y=rev.values,
    palette='viridis'
)

plt.title("Estimated Revenue by Customer Type")

plt.xlabel("Customer Type")

plt.ylabel("Estimated Revenue")

plt.xticks(rotation=20)

plt.show()
```

***Most Profitable Customer segement is the "Transient" Customer type.***

4. Average Distribution of bookings by hotel type?

```python
# Total bookings 

df['total_book']=len(df)
```

```python
avg_book=df.groupby('hotel').size().reset_index(name='TotalBookings')
avg_book
```

```python
bookings=df['hotel'].value_counts()
bookings
```

```python
plt.figure(figsize=(6,4))
bookings.plot(kind='bar')
plt.title("Distribution of Bookings by hotel type")

plt.xlabel("Hotel Type")

plt.ylabel("Number of Bookings")

plt.xticks(rotation=20)

plt.show()
```

5. Do customers requesting more special requests stay longer?

```python
relation= df.groupby('total_of_special_requests')['total_nights'].mean()

round(relation,2)
```

```python
plt.figure(figsize=(8,5))

sns.lineplot(
    x=relation.index,
    y=relation.values,
    marker='o'
)

plt.title("Average Stay vs Special Requests")

plt.xlabel("Total of Special Requests")

plt.ylabel("Total nights stay")

plt.show()
```

6. Which market segments book the farthest in advance?

```python
advance = df.groupby('market_segment')['lead_time'].mean().sort_values(ascending=False)

round(advance,2)
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x=advance.values,
    y=advance.index,
    palette='magma'
)

plt.title("Average Lead Time by Market Segment")

plt.xlabel("Lead Time (Days)")

plt.ylabel("Market Segment")

plt.show()
```
