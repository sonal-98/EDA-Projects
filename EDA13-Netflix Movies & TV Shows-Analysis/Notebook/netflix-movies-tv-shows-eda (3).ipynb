**Changing datatype of 'date_added' and 'release_year' columns from object to datetime**
 

```python
df["date_added"]=pd.to_datetime(df["date_added"],format='mixed')
df["release_year"]=pd.to_datetime(df["release_year"], format='mixed')
```

**h] Creating New Columns in the DataFrame**

<span style="color :red;">
Creating these columns allows us to find out: <br>
Which year had the most content added? <br>
Which month sees the highest number of additions?<br>
</span>

```python
# Once date_added is a datetime type, we can extract useful information.

df["year_info"] = df["date_added"].dt.year
df["month_info"] = df["date_added"].dt.month_name()
```

```python
#Adding new column named 'Status' to the DataFrame 

duration = df['date_added']- df['release_year']

df['Status']=np.where(duration.dt.days>365,"Hit on Netflix","Not a Hit")
```

**Checking for missing rows and duplicate value and treating them**

```python
#Find Missing Values

df.isnull().sum()
```

```python
#Fiiling the missing values with 'NaN'

df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")
df["country"] = df["country"].fillna("NaN")
df["rating"] = df["rating"].fillna("Rating not available")
```

```python
#For year_info, month_info, and duration columns, we can drop the rows

df = df.dropna(subset=["year_info"])
df = df.dropna(subset=["month_info"])
df = df.dropna(subset=["duration"])
```

```python
#Counts the number of duplicate rows

df.duplicated().sum()
```

```python
# Dropping the duplicates, if any

df.drop_duplicates(inplace=True)       #inplace=True updates the original DataFrame directly.
```

## Step 3: 
### Exploratory analysis by creating professional visualizations with Matplotlib and Seaborn



> ****a] Which type of content is more common on Netflix, Movies or TV Shows?****

```python
plt.figure(figsize=(6,4))           # Setting the chart size   

colors=['#E50914', '#221F1F']       # Defining colors for the Bars

# value_counts() counts occurrences of each category
df["type"].value_counts().plot(kind="bar",color=colors, edgecolor='black')      #Creating a Bar chart 

plt.title("Movies vs TV Shows on Netflix", fontsize=14, fontweight='bold')
plt.xlabel("Content Type", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.xticks(rotation=0)              # Keeps labels horizontal
plt.show()                          #plt.show() display the chart
```

*** Movie-type content dominates the Netflix platform***

> **b] Most of the Netflix content released in which year?**

```python
plt.figure(figsize=(10,5))

#using NumPy to calculate bin ranges from minimum year to maximum year
min_year=int(df["release_year"].min())
max_year=int(df["release_year"].max())
bins=np.arange(min_year, max_year+2)

#Create the histogram plot
plt.hist(df["release_year"], bins=bins, edgecolor="Black", color="Blue", alpha=0.7)

plt.title("Distribution of Netflix content by Release Years")
plt.xlabel("Release Year", fontsize=12)
plt.ylabel("Number of Titles",fontsize=12)

#Customizing x axis ticks  to show every 5 years
plt.xticks(np.arange(min_year,max_year,5), rotation=45)

plt.grid(axis="y",linestyle="--",alpha=0.7)
plt.tight_layout()
plt.show()
```

***Most netflix content released in year 2019***

> **c] Which countries produce the most Netflix content?**

```python
top_country = df["country"].value_counts().head(10)

plt.figure(figsize=(10,5))

top_country.plot(kind="bar", color="Green", edgecolor="Black")

plt.title("Top 10 Countries")
plt.xlabel("Country", fontsize=12)
plt.ylabel("Number of Titles", fontsize=12)

plt.xticks(rotation=45)
  
plt.tight_layout()          # Adjust spacing nicely, and ensures labels don't get cut off
plt.show()
```

***USA produce the most Netflix content***

> **d] Which genres are the most watched on Netflix Platform?**

```python
top_genres = df["listed_in"].value_counts().head(10)

plt.figure(figsize=(12,6))

top_genres.plot(kind="barh", color="Grey", )  #Genre names are long, that'swhy used horizontal chart to improves readability.

plt.title("Top 10 Genres")
plt.xlabel("Count", fontsize=12)
plt.ylabel("Genre", fontsize=12)

plt.tight_layout()          
plt.show()

```

***Dramas, International movies are the most watched genre***

> **e] In which year did Netflix add the highest number of content?**

```python
df['date_added']=pd.to_datetime(df['date_added'])    #convert to datetime datatype
df['date_added']=df['date_added'].dt.year              # Extracting the Year from the date

date_count= df['date_added'].value_counts().sort_index()     

plt.figure(figsize=(10,5))
sns.set_style("whitegrid")  #adding a clean background grid

plt.plot(date_count.index, date_count.values, color='#C62828', linewidth=2, marker='o', markersize=6)

plt.fill_between(date_count.index, date_count.values, color='#C62828', alpha=0.5)  #adding area shade
plt.title("Netflix content added by Years", fontsize=15, fontweight='bold')
plt.xlabel("Year", fontsize=12)
plt.ylabel("Number of content added", fontsize=12)

plt.tight_layout()          
plt.show()
 
```

> **f ] Rating distribution as the percentage of total**

```python
rating_counts= df['rating'].value_counts()
plt.figure(figsize=(8,8))
plt.pie(rating_counts.values, labels=rating_counts.index,autopct="%1.1f%%", startangle=180)  #autopct stands for automatic percentage, used to calculate percetage Labels
plt.title('Distribution of content')
plt.show()
```
